```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def compute_factors(n, Omega_d=2, Lambda=2, L=2, d=6):
    factor_t = 4 * (n + 2) * Omega_d * Lambda**(d - 4) / ((2 * np.pi)**d * L)
    factor_u = 4 * (n + 8) * Omega_d * Lambda**(d - 4) / ((2 * np.pi)**d * L)
    return factor_t, factor_u

n = 1 # i chose d,n,l and lambda and ommega randomly
factor_t, factor_u = compute_factors(n)

def beta_t_full(t, u):
    return -4 * t + factor_t * u

def beta_u_full(t, u):
    return -2 * u - factor_u * u**2


t_vals = np.linspace(-1.5, 0.5, 30)
u_vals = np.linspace(-1.0, 0.5, 30)
T, U = np.meshgrid(t_vals, u_vals)

beta_t_vals = beta_t_full(T, U)
beta_u_vals = beta_u_full(T, U)

magnitude = np.sqrt(beta_t_vals**2 + beta_u_vals**2)

beta_t_vals /= magnitude
beta_u_vals /= magnitude
```

```python
plt.figure(figsize=(8, 6))
plt.quiver(T, U, beta_t_vals, beta_u_vals, color='blue', alpha=0.7, pivot='middle', scale=20)

plt.scatter(0, 0, color='red', label="Gaussian fixed point (0, 0)", zorder=5)
plt.scatter(-1, -0.5, color='green', label="Interacting fixed point", zorder=5)

plt.axhline(0, color='black', linewidth=0.8, linestyle='--', alpha=0.7)
plt.axvline(0, color='black', linewidth=0.8, linestyle='--', alpha=0.7)
plt.title("RG Flow in the t_u Plane")
plt.xlabel("$t$")
plt.ylabel("$u$")
plt.legend()
plt.grid(alpha=0.4)
plt.show()
```
