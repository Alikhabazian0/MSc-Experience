```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def compute_sigma_tao(data, tao):
    N = len(data)
    M = int(N / tao)
    sigma_tao = np.zeros(N-tao+1)

    for i in range(M):
        x_i = data[i*tao : (i+1)*tao]
        x_i_mean = np.mean(x_i)
        sigma_i = (1/tao) * np.sum((x_i - x_i_mean)**2)
        sigma_tao += sigma_i
    
    sigma_tao /= M
    return sigma_tao
```

PART A

```python
fbm_data = np.loadtxt("C:\\Users\\user\\OneDrive\\Desktop\\FBM.txt")
tao = 10  
sigma_tao_fbm = compute_sigma_tao(fbm_data, tao)
print(sigma_tao_fbm)
```

PART B

```python
fgn_data = np.loadtxt("C:\\Users\\user\\OneDrive\\Desktop\\FGN.txt")
tao = 10
sigma_tao_fgn = compute_sigma_tao(fgn_data, tao)
print(sigma_tao_fgn)
```

PART C

```python
increment_data = np.diff(fbm_data)
tao = 10
sigma_tao_increment = compute_sigma_tao(increment_data, tao)
print(sigma_tao_increment)
```

PART D

```python
profile_data = np.cumsum(fgn_data)
tao = 10
sigma_tao_profile = compute_sigma_tao(profile_data, tao)
print(sigma_tao_profile)
```
