```python
import os
import numpy as np
import matplotlib.pyplot as plt
```

Set 2. Question 1. PART A

```python
data_file = 'C:/Users/user/OneDrive/Desktop/data.txt'
data = np.loadtxt(data_file, delimiter=',')
output_directory = 'data_sets'

if not os.path.exists(output_directory):
    os.makedirs(output_directory)
with open(data_file, 'r') as file:
    lines = file.readlines()
lines_per_file = len(lines) // 100

line_chunks = [lines[i:i+lines_per_file] for i in range(0, len(lines), lines_per_file)]

for i, chunk in enumerate(line_chunks):
    directory = os.path.join(output_directory, f'data_set_{i+1}')
    os.makedirs(directory, exist_ok=True)

    filename = os.path.join(directory, f'data_set_{i+1}.txt')
    with open(filename, 'w') as file:
        file.writelines(chunk)

    print(f'Saved {filename}')
```

Set 2. Question 1. Part B

```python
output_file = 'statistics.txt'

with open(output_file, 'w') as file:
    file.write('Data Set\tMean\tVariance\tStandard Deviation\n')

    for i, chunk in enumerate(line_chunks):
        data = np.array([float(line.strip()) for line in chunk])

        mean = np.mean(data)
        variance = np.var(data)
        std_dev = np.std(data)

        file.write(f'Data Set {i+1}\t{mean}\t{variance}\t{std_dev}\n')

        print(f'Statistics for Data Set {i+1}:')
        print(f'Mean: {mean}')
        print(f'Variance: {variance}')
        print(f'Standard Deviation: {std_dev}\n')

print(f'Statistics saved to {output_file}')
```

Set 2. Question 1. Part C

```python
data_sets = range(1, 102)
means = []
variances = []
std_devs = []

with open(output_file, 'r') as file:
    next(file)

    for line in file:
        data_set, mean, variance, std_dev = line.strip().split('\t')
        means.append(float(mean))
        variances.append(float(variance))
        std_devs.append(float(std_dev))

plt.figure(figsize=(12, 6))
plt.subplot(1, 3, 1)
plt.plot(data_sets, means, 'b.')
plt.xlabel('Data Set')
plt.ylabel('Mean')
plt.title('Mean')

plt.subplot(1, 3, 2)
plt.plot(data_sets, variances, 'r.')
plt.xlabel('Data Set')
plt.ylabel('Variance')
plt.title('Variance')

plt.subplot(1, 3, 3)
plt.plot(data_sets, std_devs, 'g.')
plt.xlabel('Data Set')
plt.ylabel('Standard Deviation')
plt.title('Standard Deviation')

plt.tight_layout()
plt.show()
```
