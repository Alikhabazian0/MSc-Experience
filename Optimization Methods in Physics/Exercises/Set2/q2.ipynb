```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_file = 'C:\\Users\\user\\OneDrive\\Desktop\\xnew.txt'
with open(x_file, 'r') as file:
    x_data = np.loadtxt(file, skiprows=1, usecols=(1, 2))

y_file = 'C:\\Users\\user\\OneDrive\\Desktop\\ynew.txt'
with open(y_file, 'r') as file:
    y_data = np.loadtxt(file, skiprows=1, usecols=(1, 2))

z_data = np.tanh(x_data[:, 0] ** 2) + np.exp(y_data[:, 0])
```

```python
plt.errorbar(x_data[:, 0], z_data, yerr=x_data[:, 1], fmt='b.', label='Series z')
plt.errorbar(x_data[:, 0], y_data[:, 0], yerr=y_data[:, 1], fmt='r.', label='Quantity y')
plt.xlabel('x')
plt.ylabel('Value')
plt.title('Series z and Quantity y')
plt.legend()
plt.grid(True)
plt.show()
```
