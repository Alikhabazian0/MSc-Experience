```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
```

Q5 - Part A


```python
def pdf(s):
    return (1/5.4)*(np.cosh(s)/((s+10)**2) + np.tanh(s))**2

s_range = np.linspace(-4, 4, 1000)
s_data = np.random.choice(s_range, size=1000, p=pdf(s_range)/np.sum(pdf(s_range)))

hist, bins = np.histogram(s_data, bins=50, density=True)
bin_centers = 0.5 * (bins[:-1] + bins[1:])
```

```python
plt.figure(figsize=(8, 6))
plt.plot(s_range, pdf(s_range), label='Theoretical PDF')
plt.hist(s_data, bins=50, density=True, label='Generated data')
plt.xlabel('s')
plt.ylabel('PDF')
plt.title('Theoretical PDF vs Generated Data PDF')
plt.legend()
plt.show()
```

Q5 - part B

```python
mean = np.mean(s_data)

def moment_integrand(s, n):
    return (s - mean)**n * pdf(s)

M3 = quad(moment_integrand, -4, 4, args=(3))[0]
M5 = quad(moment_integrand, -4, 4, args=(5))[0]


K3 = np.mean((s_data - mean)**3)
K4 = np.mean((s_data - mean)**4) - 3 * (np.mean((s_data - mean)**2))**2
K5 = np.mean((s_data - mean)**5) - 10 * np.mean((s_data - mean)**3) * np.mean((s_data - mean)**2)
```

```python
print(f"Third moment (M3): {M3:.4f}")
print(f"Fifth moment (M5): {M5:.4f}")
print(f"Third cumulant (K3): {K3:.4f}")
print(f"Fourth cumulant (K4): {K4:.4f}")
print(f"Fifth cumulant (K5): {K5:.4f}")
```
