Part A:

```python
import random
import math
```

```python

def random_walk(N):
    x = 0  
    for _ in range(N):
        step = random.choice([-1, 1])  
        x += step  
    
    return x

def compute_mean_and_variance(N, num_trials):
    sum_x = 0
    sum_x_squared = 0
    
    for _ in range(num_trials):
        x = random_walk(N)
        sum_x += x
        sum_x_squared += x**2
    
    mean = sum_x / num_trials
    variance = (sum_x_squared / num_trials) - (mean**2)
    
    return mean, variance


N = 100  
num_trials = 10000  

mean, variance = compute_mean_and_variance(N, num_trials)
print(f"Mean: {mean}")
print(f"Variance: {variance}")
```

```python
def compute_mean_displacement_and_variance(sigma, N):

    p_plus = 0.5
    p_minus = 0.5
    
    mean_s = 0 
    mean_s_squared = sigma**2
    
    mean_displacement = N * (p_plus - p_minus) * mean_s
    variance = N * (p_plus + p_minus) * mean_s_squared
    
    return mean_displacement, variance


sigmas = [0.1, 1, 10]
N = 100 

for sigma in sigmas:
    mean_displacement, variance = compute_mean_displacement_and_variance(sigma, N)
    print(f"For σ = {sigma}:")
    print(f"  <x(N)> for N = {N} steps: {mean_displacement}")
    print(f"  (σ_n)^2 for N = {N} steps: {variance}")
    print()
```

```python
def p_x(x, N):
    
    if x == 0:
        return 1.0 / (N + 1)
    elif abs(x) == N:
        return 1.0 / (2 * (N + 1))
    else:
        return 0.0

def random_walk_variance_probability(p_x, sigma_squared):

    N = int(sigma_squared)  
    mean = 0.0
    variance = 0.0

    for x in range(-N, N + 1):
        p_x_value = p_x(x, N)
        mean += x * p_x_value
        variance += x ** 2 * p_x_value

    # Check if the computed variance matches the given sigma_squared
    if variance != sigma_squared:
        return 0.0

    # Substitute the obtained N into p(x(t)) to get p(σ^2(t))
    p_sigma_squared = 0.0
    for x in range(-N, N + 1):
        p_sigma_squared += p_x(x, N)

    return p_sigma_squared
```

```python
N_max = 20
sigma_squared_values = range(N_max + 1)
probabilities = [random_walk_variance_probability(p_x, sigma_squared) for sigma_squared in sigma_squared_values]
```

```python
plt.plot(sigma_squared_values, probabilities, 'ro-')
plt.xlabel(r'$\sigma^2$')
plt.ylabel(r'$p(\sigma^2)$')
plt.title('Probability Distribution of Variance for 1D Random Walk')
plt.show()
```

part F: (for comparing position distribution and variance distribution at t=10,100,1000)

```python
def random_walk_1d(steps, num_walks):
    positions = np.zeros(num_walks)
    for _ in range(steps):
        steps = np.random.choice([-1, 1], size=num_walks)
        positions += steps
    return positions

def compute_position_distribution(steps, num_walks):
    positions = random_walk_1d(steps, num_walks)
    hist, bins = np.histogram(positions, bins=50, density=True)
    return hist, bins

def compute_variance_distribution(steps, num_walks):
    positions = random_walk_1d(steps, num_walks)
    variances = np.var(positions, axis=0)
    hist, bins = np.histogram(variances, bins=50, density=True)
    return hist, bins
```

```python
def plot_distributions(steps, num_walks, time_steps):
    for t in time_steps:
        hist_positions, bins_positions = compute_position_distribution(t, num_walks)
        hist_variances, bins_variances = compute_variance_distribution(t, num_walks)
        
        plt.figure(figsize=(12, 6))
        plt.subplot(1, 2, 1)
        plt.plot(bins_positions[:-1], hist_positions)
        plt.xlabel('Position')
        plt.ylabel('Probability')
        plt.title('Position Distribution at t={}'.format(t))
        
        plt.subplot(1, 2, 2)
        plt.plot(bins_variances[:-1], hist_variances)
        plt.xlabel('Variance')
        plt.ylabel('Probability')
        plt.title('Variance Distribution at t={}'.format(t))
        
        plt.show()
```

```python
num_walks = 1000
time_steps = [10, 100, 1000]

plot_distributions(time_steps[-1], num_walks, time_steps)
```

Q2 - Langevin particle

A to F:

```python
# Constants
mass = 1.0  
temperature = 1.0  
friction_coefficient = 1.0  
simulation_steps = 100000  
time_step = 0.01  
```

```python
# Initialize arrays
position = np.zeros(simulation_steps)
velocity = np.zeros(simulation_steps)
```

```python
# Simulation loop
for step in range(1, simulation_steps):
    # Compute random forces
    random_force = np.random.normal(0, np.sqrt(2 * mass * temperature * friction_coefficient / time_step))
    
    # Update position and velocity
    velocity[step] = velocity[step - 1] + (-friction_coefficient * velocity[step - 1] + random_force) / mass * time_step
    position[step] = position[step - 1] + velocity[step] * time_step
```

```python
# Compute quantities
avg_velocity = np.mean(velocity)
avg_velocity_squared = np.mean(velocity**2)
velocity_correlation = np.mean(velocity[:-1] * velocity[1:])
avg_position = np.mean(position)
avg_position_squared = np.mean(position**2)
position_correlation = np.mean(position[:-1] * position[1:])
velocity_prob_density = np.histogram(velocity, bins=50, density=True)

# Theoretical predictions
theoretical_avg_velocity = 0.0
theoretical_avg_velocity_squared = temperature / mass
theoretical_velocity_correlation = theoretical_avg_velocity_squared * np.exp(-friction_coefficient * time_step / mass)
theoretical_avg_position = 0.0
theoretical_avg_position_squared = theoretical_avg_velocity_squared * time_step**2
theoretical_position_correlation = theoretical_avg_position_squared * np.exp(-friction_coefficient * time_step / mass)
```

```python
print("A: <v(t)i> =", avg_velocity)
print("B: <v(t)^2> =", avg_velocity_squared)
print("C: <v(t1)v(t2)> =", velocity_correlation)
print("D: <x(t)> =", avg_position)
print("E: <x(t)^2> =", avg_position_squared)
print("F: <x(t1)x(t2)> =", position_correlation)
```

```python
print("\nComparisons with theoretical predictions:")
print("A: Difference =", avg_velocity - theoretical_avg_velocity)
print("B: Difference =", avg_velocity_squared - theoretical_avg_velocity_squared)
print("C: Difference =", velocity_correlation - theoretical_velocity_correlation)
print("D: Difference =", avg_position - theoretical_avg_position)
print("E: Difference =", avg_position_squared - theoretical_avg_position_squared)
print("F: Difference =", position_correlation - theoretical_position_correlation)
```
