```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import dblquad
from scipy.integrate import tplquad
```

```python
x_file = 'C:\\Users\\user\\OneDrive\\Desktop\\DD1.txt'
with open(x_file, 'r') as file:
    x_data = np.loadtxt(file, usecols=(1))

print(x_data)
```

part A

```python
x=x_data
t=0
def joint_pdf(x, y, t):
    
    return 0.5 * np.exp(-(x**2 + y**2))

def marginal_pdf_x(x, t):
    
    return np.exp(-x**2)

def marginal_pdf_y(y, t):
    
    return np.exp(-y**2)

def delta_tao(tao):
    
    def integrand(x, y):
        return np.abs(joint_pdf(x, y, t) - marginal_pdf_x(x, t) * marginal_pdf_y(y, t + tao))
    
    result, _ = dblquad(integrand, -np.inf, np.inf, lambda x: -np.inf, lambda x: np.inf)
    
    return result

delta = delta_tao(tao)
print(f"delta(tao) = {delta}")
```

Part B

```python
def joint_pdf(x3, tao, x1, t):
    return 0.5 * np.exp(-(x3**2 + x1**2))

def conditional_pdf_x3_given_x2(x3, tao, x2, t):
    return np.exp(-x3**2)

def conditional_pdf_x2_given_x1(x2, tao, x1, t):
    return np.exp(-x2**2)

def marginal_pdf_x1(x1, t):
    return np.exp(-x1**2)

def delta_tao(tao):
    
    def integrand(x2, x1, x3):
        
        first_integral, _ = tplquad(lambda x3, x2, x1: joint_pdf(x3, tao, x1, t) - conditional_pdf_x3_given_x2(x3, tao, x2, t + tao) * conditional_pdf_x2_given_x1(x2, tao, x1, t) * marginal_pdf_x1(x1, t),
                                   -np.inf, np.inf,
                                   lambda x1: -np.inf, lambda x1: np.inf,
                                   lambda x1, x2: -np.inf, lambda x1, x2: np.inf)
        
        
        second_integral, _ = tplquad(lambda x3, x2, x1: conditional_pdf_x3_given_x2(x3, tao, x2, t + tao) * conditional_pdf_x2_given_x1(x2, tao, x1, t) * marginal_pdf_x1(x1, t),
                                    -np.inf, np.inf,
                                    lambda x1: -np.inf, lambda x1: np.inf,
                                    lambda x1, x2: -np.inf, lambda x1, x2: np.inf)
        
        return np.abs(first_integral - second_integral)
    
    
    result, _ = tplquad(integrand, -np.inf, np.inf,
                        lambda x1: -np.inf, lambda x1: np.inf,
                        lambda x1, x2: -np.inf, lambda x1, x2: np.inf)
    
    return result

tao = 1  

delta = delta_tao(tao)
print(f"delta(tao) = {delta}")
```
