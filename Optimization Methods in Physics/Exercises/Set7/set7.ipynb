Q1 - part A

```python
import random
```

```python
def simulate_decay(lmbda, delta_t, num_iterations):
    decayed_count = 0
    for _ in range(num_iterations):
        p = lmbda * delta_t
        if random.random() < p:
            decayed_count += 1
    return decayed_count


lmbda = 0.1  # Decay rate (per unit time)
delta_t = 0.5  
num_iterations = 10000  

decayed_count = simulate_decay(lmbda, delta_t, num_iterations)
print(f"Number of decayed events: {decayed_count}")
```

Q1 - part B

```python
def simulate_decay(lmbda, t, delta_t, num_iterations):
    decayed_count = 0
    for _ in range(num_iterations):
        p = lmbda * delta_t / t
        if random.random() < p:
            decayed_count += 1
    return decayed_count


lmbda = 0.1  
t = 10.0  
delta_t = 0.5  
num_iterations = 10000  

decayed_count = simulate_decay(lmbda, t, delta_t, num_iterations)
print(f"Number of decayed events: {decayed_count}")
```

```python
import math
```

```python
def estimate_pi(num_samples):
    inside_circle = 0
    for _ in range(num_samples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x**2 + y**2 <= 1:
            inside_circle += 1
    pi_estimate = 4 * inside_circle / num_samples
    return pi_estimate

for N in [10, 100, 1000, 10000, 100000, 1000000]:
    estimated_pi = estimate_pi(N)
    print(f"For N = {N}, estimated value of pi = {estimated_pi}")
    print(f"Absolute error = {abs(estimated_pi - math.pi)}")
    print()
```

```python
import numpy as np
```

```python
def p_v(v, b_m=2):
    return np.exp(-(b_m/2)*(v**2))
```

```python
def integral_calc(lower_bound, upper_bound, N, bm=2):
    v_x, v_y, v_z = (upper_bound - lower_bound)*np.random.random_sample((3, N)) + lower_bound
    integral_v_x = ((upper_bound - lower_bound)/N)*np.sum(p_v(v_x))
    integral_v_y = ((upper_bound - lower_bound)/N)*np.sum(p_v(v_y))
    integral_v_z = ((upper_bound - lower_bound)/N)*np.sum((v_z**2)*p_v(v_z))
    integral = ((bm/(2*np.pi))**1.5) * integral_v_x * integral_v_y * integral_v_z
    return integral
```

```python
lower_bound = -1000
upper_bound = 1000
N = 1000000
bm = 2


numerical_integral = integral_calc(lower_bound, upper_bound, N)

analytical_integral = 1 / (2*np.pi**3) / (((bm/(2*np.pi))**1.5)**2)
```

```python
print(f'Analytical integration result is: {analytical_integral}')
print(f'Numerical integration result is: {round(numerical_integral, 3)}')
```

```python
def trial_wavefunction(x, alpha):

    return np.exp(-alpha * x**2)

def local_energy(x, alpha):

    return 0.5 * alpha**2 + 0.5 * x**2 * (1 - alpha**2 * x**2)
```

```python
def metropolis_sampling(x, alpha, num_samples, num_steps):
   
    samples = []
    acceptance_count = 0

    for _ in range(num_steps):
        for _ in range(num_samples):
            x_new = x + np.random.normal(0, 1)
            probability_ratio = (trial_wavefunction(x_new, alpha) / trial_wavefunction(x, alpha))**2

            if probability_ratio >= np.random.rand():
                x = x_new
                acceptance_count += 1

            samples.append(x)

    acceptance_rate = acceptance_count / (num_samples * num_steps)
    return samples, acceptance_rate
```

```python
def variational_monte_carlo(alpha, num_samples, num_steps):
    
    x = 0  
    energy_sum = 0
    energy_squared_sum = 0

    samples, acceptance_rate = metropolis_sampling(x, alpha, num_samples, num_steps)

    for sample in samples:
        energy = local_energy(sample, alpha)
        energy_sum += energy
        energy_squared_sum += energy**2

    energy_avg = energy_sum / len(samples)
    energy_variance = energy_squared_sum / len(samples) - energy_avg**2

    return energy_avg, energy_variance, acceptance_rate
```

```python
alpha = 0.5  # Variational parameter
num_samples = 10000  
num_steps = 100

ground_state_energy, energy_variance, acceptance_rate = variational_monte_carlo(alpha, num_samples, num_steps)
```

```python
print("Ground state energy:", ground_state_energy)
print("Energy variance:", energy_variance)
print("Acceptance rate:", acceptance_rate)
```
