Stochastic process - Set 3 - Question 2

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
num_plots = 5       
sample_sizes = [10, 50, 100, 500, 1000]  
```

```python
x = np.linspace(-3, 3, 1000)  
fig, axes = plt.subplots(num_plots, 1, figsize=(8, 12))

for i, ax in enumerate(axes):
    sample_size = sample_sizes[i]

    samples = np.random.uniform(0, 1, (sample_size,))

    means = np.mean(samples)

    std_dev = np.std(samples) / np.sqrt(sample_size)

    y = (1 / (std_dev * np.sqrt(2 * np.pi))) * np.exp(-(x - means) ** 2 / (2 * std_dev ** 2))

    ax.plot(x, y)
    ax.set_title(f"Sample Size: {sample_size}")
    ax.set_xlabel("Sample Mean")
    ax.set_ylabel("Density")

    plt.tight_layout()
plt.show()
```
